````python
# luckily enough, the 6 different events in Zone0 all come from different runs
seed_event_ids = np.unique(events.run)
print(seed_event_ids)

def create_hist(
    *axes
):
    return hist.Hist(
        hist.axis.IntCategory(seed_event_ids, label='Seed Event ID'),
        *axes,
        storage = hist.storage.Double()
    )
````

````python
h_tere = create_hist(
    hist.axis.Regular(100,0,10000,label='Total Ecal Rec Energy [MeV]')
)
h_tere.fill(events.run, events.ecal_total_rec_energy)
h_tere.plot()
plt.legend()
plt.show()
````

````python
h_hmpe = create_hist(
    hist.axis.Regular(100,0,100,label=r'$\max(PE_\text{Hcal})$')
)
h_hmpe.fill(events.run, events.hcal_max_pe)

h_hmpe.plot(flow='show')
plt.yscale('log')
plt.legend()
plt.show()
````

````python
create_hist(
    hist.axis.Regular(100,0,100,label='Ecal Hit RMS [mm]')
).fill(events.run, events.ecal_hit_rms).plot(flow='show')
plt.yscale('log')
plt.legend()
plt.show()
````

````python
selection_plane = create_hist(
    hist.axis.Regular(100,0,100,label=r'$\max(PE_\text{Hcal})$'),
    hist.axis.Regular(100,0,100,label='Ecal Hit RMS [mm]')
)
selection_plane.fill(events.run, events.hcal_max_pe, events.ecal_hit_rms)
````

````python
from matplotlib.patches import Rectangle

for run in selection_plane.axes[0]:
    selection_plane[hist.loc(run),...].plot(norm='log', flow='show')
    plt.title(run)
    plt.axvline(10, color='tab:red')
    plt.axhline(20, color='tab:red')
    plt.axvline(35, color='tab:red', ls='--')
    plt.show()
````

````python
selection_plane = create_hist(
    hist.axis.Regular(100,0,100,label=r'$\max(PE_\text{Hcal})$'),
    hist.axis.Regular(800,0,8000,label='Total Ecal Rec Energy [MeV]')
)
selection_plane.fill(events.run, events.hcal_max_pe, events.ecal_total_rec_energy)
````
